```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import Audio, display
from scipy.io import wavfile

#il faut 44.100 echantillons
#c'est sin(2piphit)

rate=44100
phi=440
tabx=np.linspace(0,1,rate)
taby=np.sin(2*np.pi*phi*tabx)
plt.xlim(0,5*10**(-3))
plt.plot(tabx,taby)
plt.show()

display(Audio(taby,rate=44100))

def display_signal(signal,rate,title,width):
    tabx=np.linspace(0,width,rate)
    plt.xlim(0,5*10**(-3))
    plt.plot(tabx,signal)
    plt.title(title)
    plt.show()
    display(Audio(signal,rate=rate))

display_signal(taby,rate,'la',1)


```
